Pré-requis, sur la VM Ubuntu de formation via un terminal de commande : 
copier le fichier Naufrages.csv présent dans le répertoire /home/user/Bureau/Données dans un des conteneurs lancés : 

docker cp /home/user/Bureau/Données/Naufrages.csv namenode:/home/


Autre option : en utilisant l'interface HDFS web, ajouter le fichier Naufrages.csv dans le répertoire HDFS /user/share 

```scala
%sh
hdfs dfs -put /home/Naufrages.csv /user/share/
```

On vérife que le fichier est bien présent désormais dans HDFS  : http://localhost:9870/explorer.html#/user/share


```scala
%sh
hdfs dfs -tail /user/share/Naufrages.csv
```
