```scala
%hive
USE sales_db;
DROP TABLE IF EXISTS transactions;
```

```scala
%hive
CREATE TABLE sales_db.transactions (
    transaction_id INT,
    customer_name STRING,
    amount DOUBLE,
    transaction_date STRING
)
ROW FORMAT DELIMITED
FIELDS TERMINATED BY ','
STORED AS TEXTFILE
TBLPROPERTIES ("skip.header.line.count"="1");
```

####  Format des lignes : 
####  
####  ROW FORMAT DELIMITED : Indique que les données sont structurées avec des délimiteurs
####  FIELDS TERMINATED BY ',' : Les champs (colonnes) sont séparés par des virgules (format CSV)
####  
####  Type de stockage
####  STORED AS TEXTFILE
####  
####  Les données seront stockées sous forme de fichiers texte brut
####  Autres options possibles : PARQUET, ORC, AVRO, etc.
####  
####  Propriétés de la table
####  TBLPROPERTIES ("skip.header.line.count"="1");
####  
####  TBLPROPERTIES : Permet de définir des propriétés spécifiques à la table
####  skip.header.line.count"="1" : Ignore la première ligne du fichier lors de la lecture des données
####  Très utile quand le fichier CSV contient une ligne d'en-tête avec les noms des colonnes
####  
####  Exemple concret
####  Si votre fichier CSV contient :
####  transaction_id,customer_name,amount,transaction_date
####  1,John Doe,100.50,2024-01-15
####  2,Jane Smith,250.75,2024-01-16
####  3,Bob Johnson,75.25,2024-01-17
####  
####  Avec cette définition de table :
####  
####  La première ligne (en-tête) est ignorée grâce à skip.header.line.count=1
####  
####  Les 3 lignes de données seront correctement chargées
####  Chaque virgule séparera les colonnes
####  Les données seront converties selon les types définis (INT, STRING, DOUBLE)
####  
####  Cette configuration est parfaite pour traiter des fichiers CSV standards avec en-têtes.
