```scala
%pyspark
# Remplissage du RDD avec la liste des arbres de Paris
arbres = sc.textFile("hdfs:/share/paris/les-arbres.csv")
tableau = arbres.map(lambda ligne: ligne.split(';'))

# Création des paires (hauteur, année)
paires = tableau.map(lambda champs: (champs[9], champs[15]))

# Filtrage des valeurs vides ou des en-têtes incorrects
pairesok1 = paires.filter(lambda hauteur_annee: hauteur_annee[0] != '' and hauteur_annee[0] != 'HAUTEUR (m)')

# Conversion de la hauteur en float
pairesok2 = pairesok1.map(lambda hauteur_annee: (float(hauteur_annee[0]), hauteur_annee[1]))

# Tri décroissant par hauteur
classement = pairesok2.sortByKey(ascending=False)

# Affichage du premier élément
print(classement.first())
```
